```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(6135)
x1 = np.arange(50)
y1 = np.random.randint(0, 75, 50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([100,180,240,99,220]) # it's easier to compute with arrays
```

```python
# Two plots next to each other, we need only to specify the axes
# We need to specify their positions to each other by using columns and rows >> in the figure
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2,
                               nrows=1,
                               figsize=(7,3))
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2,
                               nrows=1,
                               figsize=(7,3))
ax1.scatter(x1,y1)
ax2.bar(x2, y2)
fig.show()
```

```python
font1 = {'family': 'sans-serif', 'color': 'blue', 'size':20}
font2 = {'family': 'monospace', 'color': 'green', 'size':14}

fig, (ax1, ax2) = plt.subplots(ncols=2,
                               nrows=1,
                               figsize=(7,3)) 
ax1.scatter(x1,y1,
            marker = '+',
            color = 'indigo',
            linestyle = '--', # connecting the data points
            linewidth = 2 )
ax2.bar(x2, y2)
ax1.set_title('First plot', fontdict = font1, loc = 'left')
ax2.set_title('Second plot', fontdict = font2, loc = 'right')
fig.show()
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7,3))
ax1.scatter(x1, y1)
ax1.annotate("This is important", xy = (20, 60), xytext = (20, 70), arrowprops = dict(facecolor='black'))
ax2.bar(x2, y2)
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1','ax3'],
                                      ['ax2', 'ax3']],
                                      figsize=(7,4))
someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1, y1)
someaxes['ax1'].set_xlabel('A Big Label', fontsize=18)
someaxes['ax2'].set_xlabel('Another Label', fontsize=18)
someaxes['ax3'].set_xlabel('Yet Another Label', fontsize=18)
plt.show
```

```python
# Adjusting the layout
# 1) Tight layout - adjusting subplots, tick labels
# Constrained layout adjusts more things
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1','ax3'],
                                      ['ax2', 'ax3']],
                                      figsize=(7,4),
                                      layout = 'constrained')
someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1, y1)
someaxes['ax1'].set_xlabel('A Big Label', fontsize=18)
someaxes['ax2'].set_xlabel('Another Label', fontsize=18)
someaxes['ax3'].set_xlabel('Yet Another Label', fontsize=18)
plt.show
```

```python
# first we make our sample data
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

#defining the figure and axes (just one this time)
fig, ax = plt.subplots(figsize=(7,3))

# call both bar and plot elements to the same axes 
ax.bar(x, y1,
       color = "indigo")
ax.plot(x, y2,
        color = "red")
```

```python
# Adding error information
y2_sd = np.std(y2)
```

```python
# Plotting our line
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color= "red")
```

```python
# Customizing errorbar appearance
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color= "red")
ax.errorbar(x,
            y2,
            yerr = y2_sd,
            fmt="none", # format, line, circle markes etc.
            ecolor="indigo",
            elinewidth=4,
            capsize=6,
            capthick=4)
```

```python
from PIL import Image
import requests
from io import BytesIO

url = "https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png"

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "image/*,*/*;q=0.8",
    "Referer": "https://en.wikipedia.org/"
}

response=requests.get(url,headers=headers)
image_file=BytesIO(response.content) # help download the content from the response
image=Image.open(image_file) # Opening the images
image
```

```python
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color="red")
# Creating a container for the image
ax_image = fig.add_axes([0.1,  # X coordinate, on the figure not the axes
                         0.11, # y coordinate on the figure, not the axes
                         0.15, # image width, spanning 15% of the whole figure width
                         0.35])# image height
ax_image.imshow(image)
ax_image.axis('off') # Removing the borders of the container around the image

# Saving your figure
path = '/Users/melit/DSI/visualization/02_activities/assignments/participation'
filename = '/visualization_with_image.png'
plt.savefig(path+filename, dpi=300)
```
