```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
tips = sns.load_dataset("tips")
print(tips)
```

```python
sns.lineplot(data = tips,    # choosing the dataset
             x='total_bill', # defining the x variable
             y = 'tip')      # defining the y variable
```

```python
sns.set_style('darkgrid')
sns.lineplot(data = tips,    # choosing the dataset
             x='total_bill', # defining the x variable
             y = 'tip') 
```

```python
# Adding title and axis labels
tipgraph = sns.lineplot(data = tips,
                        x='total_bill',
                        y='tip')
tipgraph.set(title='Tips vs. Total Bill',
             xlabel='Total bill ($)',
             ylabel='Tip amount ($)')
```

```python
# Modifying aesthetic elements
fig = plt.subplots(figsize=(10,3))
tipgraph = sns.lineplot(data=tips,
                        x='total_bill',
                        y='tip',
                        color='hotpink',
                        linestyle='--',
                        linewidth = 3,
                        marker = 'o',
                        markerfacecolor='indigo'
                        )
```

```python
# Advantange of Seaborn = easier representation of multiple variables
tipgraph = sns.scatterplot(data=tips, 
                           x='total_bill',
                           y='tip',
                           style='time', # different styles/marker shapes used for data with different time variables
                           hue='day', # color is different based on the day category/variable
                           palette=['purple', 'hotpink','deepskyblue', 'yellowgreen'])
tipgraph.set(title='Tips vs. Total bill',
             xlabel = 'Total bill ($)',
             ylabel = 'Tip amount ($)')
```

```python
sns.pairplot(data=tips)
# three quantitative variables in the original dataset - comparison plot across these
# first on the top left - histograph 
```

```python
sns.pairplot(data=tips, hue='day')
# plottin also by day 
```

```python
# Relplotting
# comparisons between different levels of variables
daysplot = sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='sex', # two colors to show the data points for different sexes
    col='day', # different panels created for different days
    kind='scatter',
    col_wrap=2) # two rows
```

```python
# Plotly - to create dynamic visualizations
import plotly.graph_objects as go # go is 'graph objects'01_materials
from wordcloud import WordCloud
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted
x1 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
```

```python
graph = go.Figure()
```

```python
graph.add_trace(go.Bar(x=x1, y=y1))
```

```python
# Making our plot, updating labels
graph.update_layout(
    title="Pirate Scores",
    xaxis_title="Pirates",
    yaxis_title="Score")
```

```python
# Customization
graph = go.Figure()
graph.add_trace(go.Scatter(x=x1, y=y1, mode='markers', 
        marker=dict(
            size=15,                # point size
            color='hotpink',        # point colour
            opacity=1,              # point transparency    
            line=dict(width=5, color='purple')   
)))

graph.update_layout(
    title='Interactive Pirate Plot',
    xaxis_title='Pirates',
    yaxis_title='Scores',
    width=500, height=500
)
```

```python
# Wordclouds and venn diagrams
# Qualitative visualizations
from wordcloud import WordCloud
df = pd.read_csv(r"C:\Users\melit\DSI\visualization\02_activities\assignments\participation\movie_quotes.csv")
print(df)
```

```python
# join all our text from each row from our quote column into a string
text = " ".join(each for each in df.quote) 

# generate our wordcloud image
wordcloud = WordCloud(background_color="white",
                      colormap = 'inferno').generate(text)
# use matplotlib syntax to put our image in a figure
fig, ax = plt.subplots(figsize=(7, 3))
ax.imshow(wordcloud, # remember 'imshow' from when we added pictures to our matplotlib axes 
          interpolation='bilinear') # this line helps smooth our image
ax.axis("off")
```

```python
# Venn-diagrams
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted
A = set(['apple', 'banana', 'watermelon'])
B = set(['pumpkin', 'blueberry', 'apple', 'key lime'])
```

```python
  diagram  = venn2_unweighted([A,B],
                              set_labels=('Fruits', 'Pies'),
                              set_colors=('blue', 'red'),
                              alpha=0.5) # transparency
```

```python
# Modifying the Venn-diagram

diagram  = venn2_unweighted([A,B],
                              set_labels=('Fruits', 'Pies'),
                              set_colors=('blue', 'red'),
                              alpha=0.5) # transparency


diagram.get_label_by_id('10').set_text('\n'.join(A - B)) # Specifying the place where we want to place the label
diagram.get_label_by_id('11').set_text('\n'.join(A & B))
diagram.get_label_by_id('01').set_text('\n'.join(B - A))
# the numbers in parantheses come from documentation
# \n. >> we want to separate texts?
```
