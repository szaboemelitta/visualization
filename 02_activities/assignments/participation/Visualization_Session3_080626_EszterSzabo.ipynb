```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0, 100, 50)
y2 = np.random.randint(0, 100, 50)
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1)
ax.plot(x,y2)
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc = 'upper right', # lower right/left, center, 
    frameon = True, #add frame around the legend
    fontsize = 12, # change the font size
    ncol = 2,  # specify the number of columns
    shadow = True)
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc = 'upper right', # lower right/left, center, 
    frameon = True, #add frame around the legend
    fontsize = 12, # change the font size
    ncol = 2,  # specify the number of columns
    shadow = False)
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc = 'upper left', # Location String loc values: best' (Axes only), 'upper right', 'upper left', 'lower left', 'lower right', 'right', 'center left', 'center right', 'lower center', 'upper center', 'center', 
    bbox_to_anchor = (1,1)) # attach the legend to the 1,1 coordinate-position in the figure
    # 1,1, top right corner of the plot
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.text(10, 95, "This value is important!")
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.text(10, 95, "This value is important!",
        ha = 'center', # alignment of the text
        color = 'green', # color
        size = 14)     # size of the label
```

```python
# Test positioning
# Either linked, attached to a data point, which might change, not so useful
# https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.text.html#
# Axis-values  
# Or position relative to the figure itself
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.axis([0, 10, 0, 10]) # creating the axis-values
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.axis([0, 10, 0, 10]) 
ax.text(1,5, ". Data:(1,5)",    transform=ax.transData) # position of the data point
ax.text(0.5, 0.1, '.Axes:(0.5, 0.1)", transform=ax.transAxes') # fraction relative to the axis
ax.text(0.2, 0.2, ". Figure:(0.2, 0.2)", transform=fig.transFigure) # position relative to the figure itself
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This value is important!', xy = (10, 95), xytext = (20, 94),
            arrowprops=dict(facecolor='black')) 
# it is pointing to a specific data point (10,95), then we specify where we want to place the annotiation (20, 94)
# arrowprops is a dictionary >> we can add multiple arguments, modifying the arrow itself 
# possible styles: https://matplotlib.org/3.1.1/api/_as_gen/matplotlib.pyplot.annotate.html
        
    
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This value is important!', xy = (10, 95), xytext = (20, 94),
            arrowprops=dict(arrowstyle='wedge', color='green'))
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right') 

# locator: to remove labels and ticks
# formatter: to remove only labels

ax.yaxis.set_major_locator(plt.NullLocator())
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right') 

# locator: to remove labels and ticks
# formatter: to remove only labels

ax.xaxis.set_major_formatter(plt.NullFormatter())
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right') 

# locator: to remove labels and ticks
# formatter: to remove only labels

ax.xaxis.set_major_locator(plt.MultipleLocator(5)) # specifying a particular intervallum for the X-axis labels
```

```python
# Rotating axis labels
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')  
plt.xticks(rotation = 45, ha ='right')
# ha = horizontal alignment, specifying which side of the label is attached to the tick mark
```

```python
# plt.xlabel() >> to modify the labels, color, font, etc.

font1 = {'family': 'serif', 'color': 'indigo'}
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')  
plt.xlabel('Shiny New X Label', color = 'indigo', fontsize = 18, fontdict = font1)
```
